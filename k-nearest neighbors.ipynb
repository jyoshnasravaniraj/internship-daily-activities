### k-nearest neighbors
            K-Nearest Neighbors is a supervised machine learning algorithm used for classification and regression that          
            predicts the class 
            label or value of a new data points by identifying the K closest data points n the training dataset and applying majority 
            voting (for classification) or averaging (for regression).

### Algorithm Steps:
             step1: select the numberof neibours k.
             step2: calculate the distance between the query point and all other data points in the dataset using a 
             chosen distance metric.
             step3: Sort the distance in ascending order and select the top K-nearest neighbors.
             step4: For classification : assign the query points the class of the majority of its neighbors.
             For regression : predict the value of the query point as the average of the k-nearest neighbors.

### Types of Distance:
            1. Euclidean Distance:

            * Euclidean Distance is the straight-line distance between two points in euclidean space.
            * it is the most commonly used distance metric in KNN.
            * d= Root of (x2-x1) square + (y2-y1) square.

            2. Manhattan Distance:

            * Manhattan Distance measures the distance between two points as the sum of the absolute differences of 
             their coordinates.
            * It represents movement along grid-like paths (similar to city blocks).
            * d= |x2-x1|+|y2-y1|

            3.Minkowski Distance:

            * Minkowski distance is a general distance measure used for vector data, where the distance is calculated based on 
            the length of the vector,and the distance value is always non-negative. 
            * It is a generalized form of both Manhattan and Euclidean distances.
            * d= (|x2-x1| square +|y2-y1| square) whole power 1/2
            * if p=1, Manhattan Distance.
            * if p=0, Euclidean Distance.
